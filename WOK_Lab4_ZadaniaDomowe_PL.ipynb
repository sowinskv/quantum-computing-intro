| Pojęcie | Konwencja lub narzędzie | Na co uważać |
|---|---|---|
| Kolejność kubitów | $\lvert q_{n-1}\ldots q_1q_0\rangle$ | `q0` jest najmniej znaczącym bitem (LSB), czyli prawym znakiem etykiety. |
| Kolejność bazy | $\lvert0\ldots00\rangle,\lvert0\ldots01\rangle,\ldots$ | Indeks amplitudy jest wartością dziesiętną etykiety binarnej. |
| Iloczyn tensorowy | `np.kron(a, b)` | Wektor po lewej opisuje bardziej znaczący podukład. |
| Kolejność operacji | Najpierw $U_1$, potem $U_2$: $U_2U_1\lvert\psi\rangle$ | W kodzie bramki dodajesz w kolejności ich wykonania. |
| Symulacja stanu | `Statevector.from_instruction(qc)` | Zaczyna od $\lvert0\ldots0\rangle$; nie dodawaj pomiarów do obwodu analizowanego tą metodą. |
| Operator obwodu | `Operator(qc).data` | Używaj obwodu samych badanych bramek, bez przygotowania stanu i pomiaru. |
| Prawdopodobieństwo | $p(z)=|\alpha_z|^2$; `state.probabilities()` | Same prawdopodobieństwa nie ujawniają wszystkich faz względnych. |
| Zgodność amplitud | `np.allclose(actual, expected, atol=1e-10, rtol=0)` | Porównuje liczby zespolone, w tym znaki i fazę globalną. |
| Równoważność fizyczna stanów | `state.equiv(other)` | Ignoruje wspólną fazę globalną; nie jest testem dokładnej równości wektorów. |

a nie przez `==`. Faza względna dotyczy relacji między amplitudami;
faza globalna mnoży cały wektor przez ten sam czynnik $e^{i\theta}$.

## 🛠️ Uniwersalne funkcje pomocnicze

---

Poniższa komórka definiuje narzędzia, które możesz wykorzystać we wszystkich zadaniach.
Nie trzeba importować osobnego pliku `.py`.

| Funkcja | Zastosowanie |
|---|---|
| `pokaz_stan(stan, nazwa)` | Stan w notacji Diraca, np. `pokaz_stan(state3, r"\psi_3")`. |
| `pokaz_prawdopodobienstwa(stan)` | Tabela Markdown: stan bazowy, prawdopodobieństwo i procent; wiersze z `probabilities_dict()`. |
| `pokaz_prawdopodobienstwa(stan, kubity=[0])` | Rozkład marginalny dla wybranego kubitu, z sumowaniem po pozostałych. |
| `pokaz_macierz(macierz, nazwa)` | Macierz w LaTeX, np. `pokaz_macierz(H, "H")`. |
| `pokaz_tabele(naglowki, wiersze)` | Czytelna tabela, np. faz lub przejść. |
| `prepare_basis_state(label)` | Przygotowanie stanu bazowego z etykiety binarnej. |
| `check_state(state, expected)` | Test normalizacji i zgodności pełnych amplitud. |

Funkcje prezentacji stanów i prawdopodobieństw przyjmują `Statevector`, listę
lub tablicę amplitud, albo obwód bez pomiarów. Dla obwodu zakładają stan wejściowy
$\lvert0\ldots0\rangle$. Etykiety wyników zachowują kolejność Qiskita,
z `q0` po prawej. Stany prezentowane są z precyzją 10 miejsc, a prawdopodobieństwa
z precyzją 3 miejsc; parametry `precyzja` można zmienić.

Wyniki prawdopodobieństw pochodzą bezpośrednio z wektora stanu, nie z serii
losowych pomiarów. Renderowanie może upraszczać współczynniki do ułamków lub
pierwiastków; testy zgodności zawsze używają oryginalnych wartości numerycznych.

Najpierw wykonaj obliczenia analityczne i wpisz własny wynik oczekiwany.
Funkcje `pokaz_stan` i `pokaz_prawdopodobienstwa` służą do prezentacji;
zgodność sprawdzaj osobno przez `check_state` lub odpowiednie testy NumPy.
Wywołania podane w komentarzach komórek są podpowiedziami — usuń znak `#`
i dostosuj nazwy zmiennych dopiero po przygotowaniu swojego rozwiązania.


Tabela prawdopodobieństw pokazuje pozycje zwrócone przez `probabilities_dict()`; wyniki o dokładnie zerowym prawdopodobieństwie mogą zostać pominięte. Procenty wyświetlane są z jednym miejscem po przecinku.

```python
import html
import numpy as np
import qiskit
from IPython.display import HTML, Math, Markdown, display
from qiskit import QuantumCircuit
from qiskit.quantum_info import Operator, Statevector
from qiskit.visualization import array_to_latex

np.set_printoptions(precision=6, suppress=True)
ATOL = 1e-10


def prepare_basis_state(label):
    """Przygotuj |q_(n-1)...q_0>; prawy znak etykiety odpowiada q0."""
    if not label or any(bit not in '01' for bit in label):
        raise ValueError("Podaj niepustą etykietę binarną.")
    qc = QuantumCircuit(len(label))
    for qubit, bit in enumerate(reversed(label)):
        if bit == '1':
            qc.x(qubit)
    return qc


def check_state(state, expected):
    """Sprawdź normalizację oraz zgodność wszystkich amplitud, łącznie z fazą."""
    expected = np.asarray(expected, dtype=complex)
    assert np.isclose(np.linalg.norm(state.data), 1, atol=ATOL, rtol=0)
    assert np.allclose(state.data, expected, atol=ATOL, rtol=0)


def _jako_stan(stan):
    """Wspólna konwersja danych używana przez funkcje prezentacji."""
    if isinstance(stan, QuantumCircuit):
        stan = Statevector.from_instruction(stan)
    elif not isinstance(stan, Statevector):
        stan = Statevector(stan)
    if stan.num_qubits is None:
        raise ValueError("Podaj stan rejestru kubitów (2, 4, 8, ... amplitud).")
    return stan


def pokaz_stan(stan, nazwa=r"\psi", precyzja=10):
    """Wyświetl stan w notacji Diraca; przyjmuje wektor lub obwód bez pomiarów."""
    stan = _jako_stan(stan)
    prefix = rf"\lvert {nazwa}\rangle = " if nazwa else ""
    latex = stan.draw(
        "latex_source", convention="ket", prefix=prefix,
        decimals=precyzja, max_size=stan.dim,
    )
    display(Math(latex))


def pokaz_prawdopodobienstwa(stan, precyzja=3, *, kubity=None):
    """Wyświetl prawdopodobieństwa w bazie obliczeniowej jako tabelę Markdown.

    Przyjmuje Statevector, tablicę amplitud lub obwód bez pomiarów.
    kubity=None oznacza cały rejestr; kubity=[0] daje rozkład marginalny q0.
    Wiersze pochodzą bezpośrednio z probabilities_dict(), zgodnie z Qiskitem.
    """
    stan = _jako_stan(stan)
    if kubity is not None:
        kubity = list(kubity)
        if not kubity or len(set(kubity)) != len(kubity):
            raise ValueError("Lista kubitów musi być niepusta i bez powtórzeń.")
        if any(not isinstance(q, int) or not 0 <= q < stan.num_qubits for q in kubity):
            raise ValueError("Indeks kubitu wykracza poza rejestr lub nie jest całkowity.")

    state_header = "Stan"
    if kubity is not None:
        order = ', '.join(f'q{q}' for q in reversed(kubity))
        state_header += f" ({order})"
    rows = [
        f"| {state_header} | Prawdopodobieństwo | Procent |",
        "|:---:|---:|---:|",
    ]
    for label, probability in stan.probabilities_dict(qargs=kubity).items():
        rows.append(
            rf"| $\vert {label}\rangle$ "
            f"| {probability:.{precyzja}f} "
            f"| {100 * probability:.1f}% |"
        )
    display(Markdown("\n".join(rows)))


def pokaz_macierz(macierz, nazwa="M", precyzja=10):
    """Wyświetl macierz NumPy lub operator Qiskita jako macierz LaTeX."""
    if isinstance(macierz, Operator):
        macierz = macierz.data
    macierz = np.asarray(macierz, dtype=complex)
    if macierz.ndim != 2:
        raise ValueError("Oczekiwana jest tablica dwuwymiarowa.")
    prefix = rf"{nazwa} = " if nazwa else ""
    latex = array_to_latex(
        macierz, precision=precyzja, prefix=prefix, source=True,
        max_size=max(3, *macierz.shape),
    )
    display(Math(latex))


def pokaz_tabele(naglowki, wiersze):
    """Wyświetl czytelną tabelę HTML bez dodatkowej biblioteki tabelarycznej."""
    style = "padding:7px 12px;border-bottom:1px solid #cbd5e1;text-align:center;"
    heading = ''.join(f'<th style="{style}">{html.escape(str(v))}</th>' for v in naglowki)
    body = []
    for row in wiersze:
        if len(row) != len(naglowki):
            raise ValueError("Liczba komórek wiersza musi odpowiadać liczbie nagłówków.")
        body.append('<tr>' + ''.join(
            f'<td style="{style}">{html.escape(str(v))}</td>' for v in row
        ) + '</tr>')
    display(HTML(
        '<div style="overflow-x:auto;"><table style="border-collapse:collapse;">'
        f'<thead><tr>{heading}</tr></thead><tbody>{"".join(body)}</tbody></table></div>'
    ))


print("Qiskit:", qiskit.__version__)
```

## Przykład sposobu opracowania odpowiedzi — bramka $S$

---

**To odrębny przykład instruktażowy, a nie rozwiązanie któregokolwiek zadania.**

**Polecenie przykładowe.** Dla stanu
$\lvert\chi\rangle=\frac{\sqrt3}{2}\lvert0\rangle+\frac12\lvert1\rangle$
wyznacz wynik działania $S=\begin{bmatrix}1&0\\0&i\end{bmatrix}$,
oblicz prawdopodobieństwa pomiaru i zweryfikuj wynik w Qiskicie.

### A. Obliczenia analityczne

Przyjmujemy bazę $(\lvert0\rangle,\lvert1\rangle)$. Stan jest znormalizowany,
ponieważ $3/4+1/4=1$. Mnożenie macierzy przez wektor daje

$$
\lvert\chi'\rangle=S\lvert\chi\rangle
=\begin{bmatrix}1&0\\0&i\end{bmatrix}
\begin{bmatrix}\sqrt3/2\\1/2\end{bmatrix}
=\begin{bmatrix}\sqrt3/2\\i/2\end{bmatrix}
=\frac{\sqrt3}{2}\lvert0\rangle+\frac{i}{2}\lvert1\rangle.
$$

Z reguły Borna otrzymujemy
$P(0)=|\sqrt3/2|^2=3/4$ oraz $P(1)=|i/2|^2=1/4$.

### B. Niezależna weryfikacja w Pythonie i Qiskicie

Tablica `expected_demo` poniżej zawiera wynik wyprowadzony przed symulacją.
Kod sprawdza zarówno amplitudy, jak i prawdopodobieństwa oraz normalizację.

Do prezentacji stanu analitycznego i wyniku symulacji używamy tej samej
funkcji `pokaz_stan`, a rozkład wyników wyświetlamy przez
`pokaz_prawdopodobienstwa`. Instrukcje `assert` poniżej sprawdzają zgodność
niezależnie od sposobu wyświetlania.


```python
# Dane i przewidywanie wynikające z obliczeń analitycznych:
initial_demo = np.array([np.sqrt(3) / 2, 1 / 2], dtype=complex)
expected_demo = np.array([np.sqrt(3) / 2, 1j / 2], dtype=complex)
expected_prob_demo = np.array([3 / 4, 1 / 4])

# Przygotowanie stanu i wykonanie badanej operacji:
qc_demo = QuantumCircuit(1)
qc_demo.initialize(initial_demo, [0])
qc_demo.s(0)
state_demo = Statevector.from_instruction(qc_demo)

print(qc_demo.draw(output="text"))
pokaz_stan(state_demo, r"\chi_{\mathrm{sym}}")
pokaz_stan(expected_demo, r"\chi_{\mathrm{teoria}}")
pokaz_prawdopodobienstwa(state_demo)

assert np.isclose(np.linalg.norm(state_demo.data), 1, atol=ATOL, rtol=0)
assert np.allclose(state_demo.data, expected_demo, atol=ATOL, rtol=0)
assert np.allclose(state_demo.probabilities(), expected_prob_demo, atol=ATOL, rtol=0)
assert np.isclose(state_demo.probabilities().sum(), 1, atol=ATOL, rtol=0)
print("Wynik analityczny jest zgodny z symulacją.")
```

### Moje obliczenia analityczne — zadanie 1

**Konwencja bazy i stan początkowy:** baza $(\lvert0\rangle,\lvert1\rangle)$, $\lvert\psi_0\rangle=\lvert0\rangle=\begin{bmatrix}1\\0\end{bmatrix}$.

**Kolejne kroki obliczeń / uzupełniona tabela:**
$$
\lvert\psi_1\rangle=H\lvert0\rangle
=\frac{1}{\sqrt2}\begin{bmatrix}1&1\\1&-1\end{bmatrix}\begin{bmatrix}1\\0\end{bmatrix}
=\frac{1}{\sqrt2}\begin{bmatrix}1\\1\end{bmatrix}
=\frac{1}{\sqrt2}\lvert0\rangle+\frac{1}{\sqrt2}\lvert1\rangle=\lvert+\rangle
$$
$$
\alpha_1=\langle1\vert\psi_1\rangle=\frac{1}{\sqrt2},\qquad
P(1)=|\alpha_1|^2=\frac12

4. CNOT($q_0\to q_1$). Rozdzielamy według sterowania i korzystamy z $X\lvert-\rangle=\tfrac{1}{\sqrt2}(\lvert1\rangle-\lvert0\rangle)=-\lvert-\rangle$:
$$\lvert\psi_3\rangle=\tfrac{1}{\sqrt2}\left(\lvert-\rangle\lvert0\rangle+\lvert-\rangle\lvert1\rangle\right)
\ \xrightarrow{\mathrm{CNOT}}\
\tfrac{1}{\sqrt2}\left(\lvert-\rangle\lvert0\rangle+(X\lvert-\rangle)\lvert1\rangle\right)
=\tfrac{1}{\sqrt2}\left(\lvert-\rangle\lvert0\rangle-\lvert-\rangle\lvert1\rangle\right)$$
$$\lvert\psi_4\rangle=\lvert-\rangle\otimes\lvert-\rangle=\tfrac12(\lvert00\rangle-\lvert01\rangle-\lvert10\rangle+\lvert11\rangle)=\tfrac12[1,-1,-1,1]^T$$

Cel pozostaje w $\lvert-\rangle$, a kubit sterujący zmienia się z $\lvert+\rangle$ na $\lvert-\rangle$ — faza $-1$ wraca do sterowania.

**Wynik do porównania z symulacją:**

| Etap | Stan | Wektor |
|:---:|:---:|:---:|
| po $H(q_0)$ | $\lvert0\rangle\lvert+\rangle$ | $\tfrac{1}{\sqrt2}[1,1,0,0]^T$ |
| po $X(q_1)$ | $\lvert1\rangle\lvert+\rangle$ | $\tfrac{1}{\sqrt2}[0,0,1,1]^T$ |
| po $H(q_1)$ | $\lvert-\rangle\lvert+\rangle$ | $\tfrac12[1,1,-1,-1]^T$ |
| po CNOT | $\lvert-\rangle\lvert-\rangle$ | $\tfrac12[1,-1,-1,1]^T$ |

### Moje wnioski — zadanie 8

**Czy wyniki analityczne i numeryczne są zgodne?** Tak — obwód $H(q_0)$, $X(q_1)$, $H(q_1)$ daje dokładnie $\tfrac12[1,1,-1,-1]^T$, łącznie ze znakami.

**Odpowiedź na pytanie koncepcyjne i uzasadnienie:** $H$ na $q_0$ tworzy równą superpozycję $\lvert+\rangle$. $X$ na $q_1$ przygotowuje $\lvert1\rangle$, z którego $H$ tworzy $\lvert-\rangle$ — to daje minus przy składnikach z $q_1=1$. Bramka dwukubitowa jest zbędna, bo stan docelowy jest produktowy (niesplątany). Wszystkie prawdopodobieństwa wynoszą $\tfrac14$, tak samo jak dla $H\otimes H\lvert00\rangle=\tfrac12[1,1,1,1]^T$, choć to inny stan. Dlatego poprawność znaków potwierdza tylko porównanie amplitud.
